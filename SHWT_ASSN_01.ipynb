## Question #7
Given the below input data types in Python, convert the following hash (dictionary) into a list that all have the same denomination. e.g. if a number is "1" in thousands, turn it into 1,000.
- Python: {10: "thousands", 1000: "ones", .0005: "millions", .000000324234: "trillions"}



Example Output:
```
[10000, 1000, 500, 324234]
```


````python
val = {10: "thousands", 1000: "ones", .0005: "millions", .000000324234: "trillions"}
definitions = {
    "ones": 1,
    "thousands": 10**3,
    "millions": 10**6,
    "trillions": 10**12
}
result = []
for number, unit in val.items():
    multiplier = definitions[unit]
    res = number * multiplier
    res = int(res)
    result.append(res)
print(result)
````
