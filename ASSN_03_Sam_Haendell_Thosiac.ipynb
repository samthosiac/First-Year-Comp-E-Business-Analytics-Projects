# 5. What is the average departure delay time, for each airline? Who has the lowest? The highest? (hint.. group_by functions should be useful)

The average list of each airlines will be found below but the lowest average was for the US airline while the highest was for the F9 airline.

Python

```python
# Average departure delay by airline
avg_dep_delay = flights.groupby("carrier", as_index=False)["dep_delay"].mean()

avg_dep_delay.columns = ["carrier", "avg_dep_delay"]

avg_dep_delay
```

```python
lowest = avg_dep_delay.loc[avg_dep_delay["avg_dep_delay"].idxmin()]
lowest
```

```python
highest = avg_dep_delay.loc[avg_dep_delay["avg_dep_delay"].idxmax()]
highest
```

R

```python

%%R
avg_dep_delay <- flights %>%
  group_by(carrier) %>%
  summarise(avg_dep_delay = mean(dep_delay, na.rm = TRUE))

avg_dep_delay

avg_dep_delay %>% slice_min(avg_dep_delay)

avg_dep_delay %>% slice_max(avg_dep_delay)
```

# 6., 7., & 8. Create new dataframes, where you...
1. filter the dataset to just American Airlines, Delta, and United Airlines
2. Aggregate the dataset so that only the airline field and year are your only columns, and the count of flights in a given year are the values
3. Widen the dataset so that each airline is its own column

Python

```python
filtered = flights[flights["carrier"].isin(["AA", "DL", "UA"])]
filtered
```

```python
agg = (
    filtered
    .groupby(["carrier", "year"], as_index=False)
    .size()
    .rename(columns={"size": "flight_count"})
)
agg
```

```python
wide = agg.pivot(index="year", columns="carrier", values="flight_count").reset_index()
wide
```

R

```python
%%R
filtered <- flights %>%
  filter(carrier %in% c("AA", "DL", "UA"))
filtered
```

```python
agg <- filtered %>%
  group_by(carrier, year) %>%
  summarise(flight_count = n(), .groups = "drop")
agg
```

```python
wide <- agg %>%
  pivot_wider(names_from = carrier, values_from = flight_count)

wide
```

# 9. Clean up the column names of the original flights dataset, to make it more human readable
