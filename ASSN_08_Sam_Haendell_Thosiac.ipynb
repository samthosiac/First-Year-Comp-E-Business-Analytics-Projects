Download Data, Read it and Find the latest Period while savinf it in a new colunm called period ending.


Period ending goes over all the data and report us with the single most recent date whcih does nit permit much comparison. Due to that I have decided to also get the most recent year apart from that because it wil give me more data and match what I am looking for more effectively.

```python
import pandas as pd

fund = pd.read_csv("fundamentals.csv")
fund["Period Ending"] = pd.to_datetime(fund["Period Ending"])

latest_period = fund["Period Ending"].max()
df = fund[fund["Period Ending"] == latest_period].copy()
df
```

Getting the latest year variables and making a copy of tis data frame as well.

```python
latest_year = fund["For Year"].max()
df = fund[fund["For Year"] == latest_year].copy()
df
```

There I am creating a better table taht only contains the different features of the dataset in fundamentals.csv while dropping any NA columns. I am also adding the  Ticker Symbol just for them to be referenced through the table I am creating.

```python
features = [
    "Total Revenue",
    "Gross Profit",
    "Operating Income",
    "Net Income",
    "Total Assets",
    "Total Liabilities",
    "Operating Margin",
]

df_feat = df[["Ticker Symbol"] + features].dropna()
df_feat
```

I am now importing my tools (function containers) to permit the PCA. The Stadnard Scaler will bring all the variable to the same scale and let us compare them while the PCA will compress them into principal components.





```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = df_feat[features].values
X_scaled = StandardScaler().fit_transform(X)

pca = PCA()
X_pca_full = pca.fit_transform(X_scaled)
X_pca_full
```

```python
X_scaled
```

Here I plotted how much total variance is explained as the model added more principal components.
pca.explained_variance_ratio_ is a list containing those usual variances:
[0.55, 0.25, 0.10, 0.05, 0.03, 0.02]
While np.cumsum() utilizes numpy to turned the value mentioned above into cumulative variances.

```python
import matplotlib.pyplot as plt
import numpy as np

cum_var = np.cumsum(pca.explained_variance_ratio_)

plt.plot(range(1, len(cum_var)+1), cum_var, marker="<")
plt.axhline(0.9, color="red", linestyle="dotted")
plt.title("Cumulative Explained Variance")
plt.xlabel("Components")
plt.ylabel("Cumulative Variance")
plt.show()
```

```python
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
```

From the line above, the model will create a PCA model with 3 principal components and then transforms the scaled data to 3 components.

This code line below utilizes KMeans to test our different number of components in clusters just for us to be able to see what works or not and what the best option is.

```python
from sklearn.cluster import KMeans

inertias = []
num = range(2, 11)

for k in num:
    km = KMeans(n_clusters=k, random_state=35, n_init=10)
    km.fit(X_pca)
    inertias.append(km.inertia_)

plt.plot(num, inertias, marker="d")
plt.title("Elbow Plot")
plt.xlabel("k")
plt.ylabel("Inertia")
plt.show()
```

This line of text below deos the same as the one above but it wuses the Silhouette Score instead.

```python
from sklearn.metrics import silhouette_score

sil_scores = []

for k in num:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca)
    sil_scores.append(silhouette_score(X_pca, labels))

plt.plot(num, sil_scores, marker="o")
plt.title("Silhouette Scores")
plt.xlabel("num")
plt.ylabel("Silhouette")
plt.show()
```

After seen both graphs, I havde decided to take 4 clusters as the optimal number. As I haev seen here inertia lowers from 2 to 4 but once it goes past 4 it starts to flatten which has demanded me that I choose it as my variable.

This line of code runs K‑Means with 4 clusters and then adds the resulting cluster label to each stock in the dataframe

```python
kmeans = KMeans(n_clusters=4, random_state=35, n_init=10)
df_feat["cluster"] = kmeans.fit_predict(X_pca)
df_feat
```

There I am reading the Securities file and selecting some of their best components and merge it to our feat table which will utiize the left merge method at the end to keep our datta together as Ticker Symbol was our common point.

```python
sec = pd.read_csv("securities.csv")
df_merged = df_feat.merge(
    sec[["Ticker symbol", "GICS Sector", "GICS Sub Industry"]],
    left_on="Ticker Symbol",
    right_on="Ticker symbol",
    how="left"
)
df_merged
```

This plot has the different stocks in the 2‑D PCA space (PC1 vs PC2) and colors each point by its assigned cluster so I can inspect teh model separate and see how each cluster is behaving.

```python
import seaborn as sns

pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2", "PC3"])
pca_df["cluster"] = df_feat["cluster"]

sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="cluster", palette="Set1")
plt.title("Clusters in PCA level")
plt.show()
```

This cluster summary generates the average value of each financial feature for each cluster and gives us a heatmap visual.

```python
cluster_summary = df_feat.groupby("cluster")[features].mean().T
sns.heatmap(cluster_summary, annot=True, cmap="coolwarm")
plt.title("Cluster Feature Profiles")
plt.show()
```

This code section counts each sector composition and list their impact in the different clusters they are apart.


```python
sector_counts = df_merged.groupby("cluster")["GICS Sector"].value_counts(normalize=True)
sector_counts
```
