## Question 10

Create a function that takes an atomic vector as input. The function should return TRUE if the atomic vector contains at least 1 even number, and false otherwise. The function should be assigned to name `contains_even`. Once you create the function use the function to determine if the following atomic vector (or list in Python) contains any even numbers - `c(11,13,15,16,19)`

R

````python
%%R
contains_even <- function(x) {
  for (i in x) {
    if (i %% 2 == 0) {
      return(TRUE)
    }
  }
  return(FALSE)
}

print(contains_even(c(11, 13, 15, 16, 19)))
print(contains_even(c(11, 13, 15,19)))
````

````python
#--- Dont change this!
print(contains_even(list(11,13,15,16,19)))
print(contains_even(list(11,13,15,19)))
````

Python

````python
def contains_even(x):
  for i in x:
    if i % 2 == 0:
      return TRUE
    return False
print(contains_even([11,13,15,16,19]))
print(contains_even([11,13,15,19]))
````

## Question 11

Create a function named `counter` that given a integer and vector, will return how many times the integer appears inside the vector. The integer input should default to 10.

R

````python
%%R
counter <- function(x,y) {
  count <- 0
  for (z in y) {
    if (z == x) {
      count <- count + 1
    }
  }
  return(count)
}
print(counter(11, list(11,13,15,16,11) ))
````

````python
#--- Dont change this!
print(counter(11, list(11,13,15,16,11) ))
````

## Question 13

Without using any built-in functions, write a function named `reverse_reverse` that takes an atomic vector as an input, and returns the same list, in reverse order

R

````python
%%R
reverse_reverse <- function(x) {
  result <- c()
  for (y in seq(length(x), 1, -1)) {
    result <- c(result, y)
  }
  return(result)
}

print(reverse_reverse(c(1, 2, 3, 4, 5)))
````

````python
#--- Dont change this!
print(reverse_reverse( c(1,2,3,4,5) ))
````

Python

````python
def reverse_reverse(x):
    list = []
    for y in range(len(x),0,-1):
        list.append(y)
    return list
print(reverse_reverse( [1,2,3,4,5] ))
````

````python
#--- Dont change this!
print(reverse_reverse( [1,2,3,4,5] ))
````

## Question 14

5. Write a function named `palindromer` that takes a atomic vector of characters as an input, and returns TRUE only if it is a Palindrome, and FALSE otherwise
e.g. c("H", "A", "N", "N", "A", "H") should be TRUE

e.g. c("B", "E", "N") should be FALSE

R

````python
%%R
palindromer <- function(x) {
  reversed <- rev(x)
  return(identical(x, reversed))
}
print( palindromer( c("H", "A", "N", "N", "A", "H") ))
print( palindromer( c("B", "E", "N") ))
````

````python
#--- Dont change this!
print( palindromer( c("H", "A", "N", "N", "A", "H") ))
print( palindromer( c("B", "E", "N") ))
````

Python

````python
def palindromer(x):
    list = []
    for i in range(len(x),0,-1):
        list.append(x[i-1])
    return list == x
print( palindromer( ["H", "A", "N", "N", "A", "H"] ))
print( palindromer( ["B", "E", "N"] ))
````
