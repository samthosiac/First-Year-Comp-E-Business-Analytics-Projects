Data Cleaning and Cleansing

```r
intakes  <- clean_names(intakes)
outcomes <- clean_names(outcomes)
map      <- clean_names(map)
```

Merging The Datasets

```r
#--Merge intakes + outcomes
merged <- intakes %>%
  left_join(outcomes, by = "animal_id")

#--Merge with stray map
merged_all <- merged %>%
  left_join(map, by = "animal_id")
```

Cleaning the Dataset to take off Mostly NA Columns

```r
View(merged_all)
```

Clean by Empty Columns

```r
merged_clean <- merged_all %>%
  select(where(~ mean(is.na(.)) < 0.5))
View(merged_clean)
```

Clean The Merge By Empty Rows | Final Cleanse

```r
merged_noempty <- merged_clean %>%
  filter(!if_all(everything(), is.na))
View(merged_noempty)
```
